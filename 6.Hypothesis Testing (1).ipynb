Scenario: Investigating if weekly operating costs for Bombay Hospitality Ltd. are higher than the theoretical model suggests.

### 1. State the Hypotheses

Theoretical Mean ($\mu$): Calculated from $W = 1000 + 5X$, where $X=600$161616.$\mu = 1000 + 5(600) = 4000$.

Null Hypothesis ($H_0$): $\mu \leq 4000$ (Costs are not higher than the model)18181818.

Alternative Hypothesis ($H_1$): $\mu > 4000$ (Costs are significantly higher than the model)19191919.

### 2. Calculate the Test Statistic:

20Sample Mean ($\bar{x}$): 3,050.
   
Theoretical Mean ($\mu$): 4,000.

Standard Deviation ($\sigma$): $5 \times 25 = 125$.

Sample Size ($n$): 25.

Formula: $Z = \frac{\bar{x} - \mu}{\sigma / \sqrt{n}}$ 25$Z = \frac{3050 - 4000}{125 / \sqrt{25}} = \frac{-950}{25}$Test 

Statistic ($Z$): -38.0

### 3. Determine the Probability and Compare:

Alpha Level ($\alpha$): 0.05

Critical Value ($z_{critical}$): For a one-tailed test at 5%, the critical value is 1.645.

Decision Rule: Reject $H_0$ if $z > 1.645$.

### 4. Make a Decision 

Comparison: The test statistic (-38.0) is much smaller than the critical value (1.645) and falls deep into the non-rejection region.


Decision: Fail to reject the null hypothesis.

### 5.Conclusion 
Based on the sample data provided, there is no evidence to support the claim that weekly operating costs are higher than the model suggests. In fact, the sample mean (3,050) is significantly lower than the model's prediction (4,000).

```python
import numpy as np
from scipy import stats

# Data
sample_mean = 3050
mu = 1000 + (5 * 600)
sigma = 125
n = 25
alpha = 0.05

# Z-test calculation
z_score = (sample_mean - mu) / (sigma / np.sqrt(n))
z_critical = stats.norm.ppf(1 - alpha)

print(f"Test Statistic (z): {z_score}")
print(f"Critical Value: {z_critical}")

if z_score > z_critical:
    print("Decision: Reject the Null Hypothesis")
else:
    print("Decision: Fail to Reject the Null Hypothesis")
```

Data Overview Sample Data (millions of characters): 1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29

Sample Size ($n$): 15Sample Mean 

($\bar{x}$): 1.2387

Sample Standard Deviation ($s$): 0.1932

### a. 99% Confidence Interval Using Sample Standard Deviation

#### Rationale for using the t-distribution

We use the t-distribution here because: 

1. The population standard deviation ($\sigma$) is unknown

3. The sample size is small ($n < 30$).

##### Steps and Calculation

1.Degrees of Freedom ($df$): $n - 1 = 15 - 1 = 14$

2.Confidence Level: 99% ($\alpha = 0.01$)

3.Critical Value ($t^*$): For $df=14$ and $\alpha/2=0.005$, $t^* \approx 2.977$.

4.Standard Error ($SE$): $s / \sqrt{n} = 0.1932 / \sqrt{15} \approx 0.0499$

5.Margin of Error: $t^* \times SE = 2.977 \times 0.0499 \approx 0.1485$


##### Confidence Interval:

$$1.2387 \pm 0.1485 \rightarrow \mathbf{(1.0902, 1.3871)}$$

### b. 99% Confidence Interval Using Known Population Standard Deviation

#### Rationale for using z-distribution

We use the z-distribution (normal distribution) here because the population standard deviation ($\sigma$) is known (given as 0.2 million characters).

Steps and Calculation

1.Known $\sigma$: 0.2

2.Confidence Level: 99% ($\alpha = 0.01$)

3.Critical Value ($z^*$): For a 99% confidence level, $z^* \approx 2.576$.

5.Standard Error ($SE$): $\sigma / \sqrt{n} = 0.2 / \sqrt{15} \approx 0.0516$

Margin of Error: $z^* \times SE = 2.576 \times 0.0516 \approx 0.1330$

Confidence Interval:$$1.2387 \pm 0.1330 \rightarrow \mathbf{(1.1057, 1.3717)}$$

```python
import numpy as np
from scipy import stats

data = [1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29]
n = len(data)
mean = np.mean(data)
s = np.std(data, ddof=1)

# Task a: t-distribution
ci_t = stats.t.interval(0.99, n-1, loc=mean, scale=s/np.sqrt(n))

# Task b: z-distribution (sigma=0.2)
ci_z = stats.norm.interval(0.99, loc=mean, scale=0.2/np.sqrt(n))

print(f"99% CI (t-dist): {ci_t}")
print(f"99% CI (z-dist): {ci_z}")
```
