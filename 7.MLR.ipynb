### Task 5.Apply Lasso and Ridge methods on the model.

1. Lasso Regression (L1 Regularization)
Lasso adds a penalty equivalent to the absolute value of the magnitude of coefficients. This can shrink some coefficients to exactly zero, effectively performing feature selection.

2. Ridge Regression (L2 Regularization)
Ridge adds a penalty equivalent to the square of the magnitude of coefficients. This shrinks the coefficients of correlated features, making the model more robust without necessarily removing any variables.

##### 1.What is Normalization & Standardization and how is it helpful?
Normalization (Min-Max Scaling): Scales data to a fixed range, usually 0 to 1. It is useful when you don't know the distribution of data or when it's not Gaussian.

Standardization (Z-score Scaling): Rescales data to have a mean of 0 and a standard deviation of 1. It is more robust to outliers and is preferred for algorithms that assume a Gaussian distribution (like Linear Regression).
