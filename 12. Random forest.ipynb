### 4: Random Forest Model Implementation
### 5: Bagging and Boosting Methods


```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier, BaggingClassifier, AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report

# 1. Split
X = df_clean.drop('Type', axis=1)
y = df_clean['Type']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

# 2. Scaling
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Helper function to print results
def evaluate_model(y_true, y_pred, model_name):
    print(f"\n--- {model_name} Results ---")
    print(f"Accuracy: {accuracy_score(y_true, y_pred):.4f}")
    # Use macro average for multi-class classification
    print(f"Precision (Macro): {precision_score(y_true, y_pred, average='macro', zero_division=0):.4f}")
    print(f"Recall (Macro): {recall_score(y_true, y_pred, average='macro', zero_division=0):.4f}")
    print(f"F1-Score (Macro): {f1_score(y_true, y_pred, average='macro', zero_division=0):.4f}")

# 3. Random Forest
rf = RandomForestClassifier(n_estimators=100, random_state=42)
rf.fit(X_train_scaled, y_train)
y_pred_rf = rf.predict(X_test_scaled)
evaluate_model(y_test, y_pred_rf, "Random Forest")

# 4. Bagging
bagging = BaggingClassifier(estimator=DecisionTreeClassifier(), n_estimators=100, random_state=42)
bagging.fit(X_train_scaled, y_train)
y_pred_bagging = bagging.predict(X_test_scaled)
evaluate_model(y_test, y_pred_bagging, "Bagging")

# 5. Boosting
boosting = AdaBoostClassifier(estimator=DecisionTreeClassifier(max_depth=3), n_estimators=100, random_state=42)
boosting.fit(X_train_scaled, y_train)
y_pred_boosting = boosting.predict(X_test_scaled)
evaluate_model(y_test, y_pred_boosting, "Boosting (AdaBoost)")
```

### Additional Notes:
##### 1. Explain Bagging and Boosting methods. How is it different from each other.

Both Bagging and Boosting are "Ensemble Learning" techniques. They improve the performance of machine learning models by combining the predictions of multiple individual models (often called "weak learners").

Bagging (Bootstrap Aggregating)
How it works: Bagging builds multiple models in parallel and independently of each other. It creates "bootstrap" samples of the training data (random subsets created with replacement) and trains a separate base model on each subset. The final output is usually an average (for regression) or a majority vote (for classification) of all models.

Primary Goal: To reduce variance (i.e., prevent overfitting). It makes the model more robust by averaging out individual errors.

Example: Random Forest.

Boosting
How it works: Boosting builds models sequentially. Each new model is trained to specifically focus on the errors (misclassified instances) made by the previous model. It effectively "boosts" the performance of the weak learners by focusing on the data points that were difficult to predict.

Primary Goal: To reduce bias (i.e., improve accuracy). It turns weak learners into a strong, highly accurate model.

Example: AdaBoost, Gradient Boosting, XGBoost.

##### 2. Explain how to handle an imbalance in the data.
Imbalance occurs when one class (the majority) significantly outnumbers the other (the minority), causing the model to ignore the minority class because it isn't "economical" for the model to classify those few cases correctly.

Techniques to Handle Imbalance:
Resampling Methods:

Oversampling (SMOTE): Instead of just duplicating the minority class, SMOTE (Synthetic Minority Over-sampling Technique) creates synthetic new examples that are similar to the existing minority cases.

Undersampling: Randomly removing samples from the majority class to balance the ratio. (Warning: This can lead to losing valuable data).

Algorithm-Level Adjustments (Class Weights):

Most algorithms (like Random Forest or SVM) have a parameter called class_weight='balanced'. This tells the algorithm to punish mistakes on the minority class more severely than mistakes on the majority class, forcing the model to pay more attention to the minority group.

Use Correct Metrics:

Never use Accuracy for imbalanced data. A model can achieve 99% accuracy by simply predicting the majority class every time, failing to identify any of the minority cases.

Use F1-Score, Precision, Recall, and AUC-ROC. These metrics explicitly account for how well the model identifies the minority class (Positive class).
